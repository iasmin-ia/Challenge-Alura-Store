### Importação dos dados



```python
import pandas as pd

url = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_1.csv"
url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_2.csv"
url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_3.csv"
url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_4.csv"

loja = pd.read_csv(url)
loja2 = pd.read_csv(url2)
loja3 = pd.read_csv(url3)
loja4 = pd.read_csv(url4)

loja.head()
```

#1. Análise do faturamento



```python
# Calculando o faturamento total de cada loja a partir da soma dos valores da coluna "Preço".

faturamento_loja1 = loja['Preço'].sum()
faturamento_loja2 = loja2['Preço'].sum()
faturamento_loja3 = loja3['Preço'].sum()
faturamento_loja4 = loja4['Preço'].sum()
```

```python
# Criando um dicionário

faturamento = {
    'Loja 1': faturamento_loja1,
    'Loja 2': faturamento_loja2,
    'Loja 3': faturamento_loja3,
    'Loja 4': faturamento_loja4
}
```

```python
import pandas as pd

# Em seguida, os resultados foram organizados em uma estrutura do tipo Series do

faturamento_series = pd.Series(faturamento)
```

```python
# Aqui os valores foram formatados em reais (R$), tornando a apresentação dos

faturamento_formatado = faturamento_series.map(lambda x: f"R$ {x:,.2f}")
faturamento_formatado
```

# Análise de Desempenho Geográfico

```python
# Visualizando as colunas de latitude e longitude para entender como os dados estão organizados
loja[['lat', 'lon']].head()
```

```python
# Criando uma coluna para identificar de qual loja é cada venda
import pandas as pd
import matplotlib.pyplot as plt

loja['Loja'] = 'Loja 1'
loja2['Loja'] = 'Loja 2'
loja3['Loja'] = 'Loja 3'
loja4['Loja'] = 'Loja 4'


# Juntando os dados das quatro lojas em um único DataFrame
dados_geo = pd.concat([loja, loja2, loja3, loja4])
```

```python
# Criando um gráfico de dispersão com todas as vendas
plt.figure(figsize=(8,6))

plt.scatter(dados_geo['lon'], dados_geo['lat'], alpha=0.3)

plt.title('Distribuição Geográfica das Vendas')
plt.xlabel('Longitude')
plt.ylabel('Latitude')

plt.show()
```

```python
# Criando um gráfico de dispersão separado por loja
plt.figure(figsize=(8,6))

cores = {
    'Loja 1': 'blue',
    'Loja 2': 'orange',
    'Loja 3': 'green',
    'Loja 4': 'red'
}

for loja_nome in dados_geo['Loja'].unique():
    dados_loja = dados_geo[dados_geo['Loja'] == loja_nome]

    plt.scatter(
        dados_loja['lon'],
        dados_loja['lat'],
        alpha=0.5,
        s=15,  # tamanho do ponto
        color=cores[loja_nome],
        label=loja_nome
    )

plt.title('Distribuição Geográfica das Vendas por Loja')
plt.xlabel('Longitude')
plt.ylabel('Latitude')
plt.legend()
plt.show()
```

```python
# Criando um mapa de calor para identificar regiões com maior concentração de vendas
import seaborn as sns

plt.figure(figsize=(8,6))

sns.kdeplot(
    x=dados_geo['lon'],
    y=dados_geo['lat'],
    cmap="Reds",
    fill=True,
    thresh=0.05
)

plt.title('Mapa de Calor das Vendas')
plt.xlabel('Longitude')
plt.ylabel('Latitude')

plt.show()
```

# 📄 Relatório Final – Análise das Lojas da Alura Store

# 1 . Introdução

O objetivo desta análise foi avaliar o desempenho das quatro lojas da rede Alura Store a partir de indicadores de vendas, faturamento, avaliações dos clientes, categorias de produtos, produtos mais e menos vendidos e frete médio. A partir desses dados, buscou-se identificar a loja com menor desempenho geral, a fim de recomendar qual unidade deve ser vendida para o início de um novo empreendimento.


# 2 . Análise dos Resultados 2.1 Faturamento das lojas

A análise do faturamento total mostrou diferenças relevantes entre as unidades:

Loja 1: R$ 1.534.509,12

Loja 2: R$ 1.488.459,06

Loja 3: R$ 1.464.025,03

Loja 4: R$ 1.384.497,58

A Loja 1 apresentou o maior faturamento, enquanto a Loja 4 registrou o menor valor, indicando menor desempenho financeiro em relação às demais.


# 2.2 Vendas por categoria de produtos

Ao analisar a quantidade de vendas por categoria, observou-se que algumas categorias concentram maior volume de vendas em todas as lojas, como eletrônicos e móveis. Entretanto, a Loja 4 apresentou menor volume de vendas em diversas categorias, além de menor destaque em categorias estratégicas.

Esse comportamento sugere menor diversidade e força comercial da Loja 4 em relação às outras lojas.

# 2.3 Avaliações dos clientes

A média das avaliações dos clientes mostrou que as Lojas 1, 2 e 3 apresentam desempenho mais consistente em termos de satisfação dos consumidores. A Loja 4, por sua vez, apresentou média inferior, indicando menor percepção de qualidade por parte dos clientes.

Esse fator é relevante, pois a satisfação do cliente impacta diretamente a fidelização e o potencial de vendas futuras.


# 2.4 Produtos mais e menos vendidos

A análise dos produtos mais vendidos reforçou o padrão observado nas categorias: a Loja 4 possui menor volume de vendas dos produtos de maior destaque. Além disso, a presença de produtos com baixo desempenho foi mais evidente nessa loja, indicando menor eficiência na estratégia de vendas e mix de produtos.

# 2.5 Frete médio por loja

O frete médio das lojas também foi analisado. Apesar de o custo de frete ser pago pelo cliente, ele influencia a competitividade da loja. A Loja 4 apresentou um frete médio relativamente elevado em comparação ao seu desempenho de vendas, o que pode impactar negativamente a decisão de compra dos consumidores.

# 3. Conclusão e Recomendação

Com base nas análises realizadas, a Loja 4 se destaca como a unidade com menor desempenho geral. Ela apresentou:

*   o menor faturamento total;
*   menor volume de vendas em categorias estratégicas
*   média de avaliações inferior as demias lojas
*   menor destaque nos produtos mais vendidos
*   relação desfavorável entre desempenho de vendas e frete médio.

Diante desses fatores, recomenda-se que o Senhor João venda a Loja 4, pois ela apresenta menor eficiência comercial e menor potencial de crescimento em comparação às outras unidades da rede.
